CSV File

CSV is useful for storing data in rows and columns.

```python
import csv

students = [
    ["ID", "Name", "Age"],
    [1, "Ali", 20],
    [2, "Sara", 21]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(students)

with open("students.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

Exception Handling

Exception handling is used to handle errors without stopping the program.

try contains the code.
except handles the error.
finally runs at the end.

```python
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print(result)

except ValueError:
    print("Please enter a valid number")

except ZeroDivisionError:
    print("Cannot divide by zero")

finally:
    print("Program finished")
```

Task 10

Student Record Management System

This project stores student records in a CSV file.

CRUD means:

Create
Read
Update
Delete

```python
import csv
import os

FILE_NAME = "student_records.csv"

def create_file():
    if not os.path.exists(FILE_NAME):
        with open(FILE_NAME, "w", newline="") as file:
            writer = csv.writer(file)
            writer.writerow(["id", "name", "age", "course"])

def add_student(student_id, name, age, course):
    with open(FILE_NAME, "a", newline="") as file:
        writer = csv.writer(file)
        writer.writerow([student_id, name, age, course])

def view_students():
    with open(FILE_NAME, "r") as file:
        reader = csv.DictReader(file)

        for student in reader:
            print(student)

def update_student(student_id, name, age, course):
    with open(FILE_NAME, "r") as file:
        students = list(csv.DictReader(file))

    found = False

    for student in students:
        if student["id"] == str(student_id):
            student["name"] = name
            student["age"] = str(age)
            student["course"] = course
            found = True

    with open(FILE_NAME, "w", newline="") as file:
        writer = csv.DictWriter(
            file,
            fieldnames=["id", "name", "age", "course"]
        )
        writer.writeheader()
        writer.writerows(students)

    if found:
        print("Student updated")
    else:
        print("Student not found")

def delete_student(student_id):
    with open(FILE_NAME, "r") as file:
        students = list(csv.DictReader(file))

    new_students = []

    for student in students:
        if student["id"] != str(student_id):
            new_students.append(student)

    with open(FILE_NAME, "w", newline="") as file:
        writer = csv.DictWriter(
            file,
            fieldnames=["id", "name", "age", "course"]
        )
        writer.writeheader()
        writer.writerows(new_students)

    if len(new_students) < len(students):
        print("Student deleted")
    else:
        print("Student not found")

create_file()

add_student(1, "Ali", 20, "Python")
add_student(2, "Sara", 21, "Python")

print("Student Records")
view_students()

update_student(1, "Ali Khan", 21, "Python")

delete_student(2)

print("Updated Records")
view_students()
```

Task 11

Student Record Management System Using JSON

The same CRUD idea can also be used with a JSON file.

```python
import json
import os

FILE_NAME = "students.json"

def load_students():
    if not os.path.exists(FILE_NAME):
        return []

    with open(FILE_NAME, "r") as file:
        return json.load(file)

def save_students(students):
    with open(FILE_NAME, "w") as file:
        json.dump(students, file, indent=4)

def add_student(student_id, name, age, course):
    students = load_students()

    students.append({
        "id": student_id,
        "name": name,
        "age": age,
        "course": course
    })

    save_students(students)

def view_students():
    students = load_students()

    for student in students:
        print(student)

def update_student(student_id, name, age, course):
    students = load_students()

    for student in students:
        if student["id"] == student_id:
            student["name"] = name
            student["age"] = age
            student["course"] = course

    save_students(students)

def delete_student(student_id):
    students = load_students()

    students = [
        student for student in students
        if student["id"] != student_id
    ]

    save_students(students)

add_student(1, "Ali", 20, "Python")
add_student(2, "Sara", 21, "Python")

print("Student Records")
view_students()

update_student(1, "Ali Khan", 21, "Python")
delete_student(2)

print("Updated Records")
view_students()
```
