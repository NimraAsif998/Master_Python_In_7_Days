## 6️ Conditional Structures
We use `if`, `elif`, and `else` to make decisions. In Python, **indentation** (spaces) is very important.

````python
marks = 78

if marks >= 90:
    print("Grade: A+")
elif marks >= 80:
    print("Grade: A")
elif marks >= 70:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
else:
    print("Grade: Fail")
````

````python
# Even / Odd
number = 7
if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")

# Nested if
age = 20
has_id = True
if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID is required")
else:
    print("You are under 18")

# Ternary (one-line if)
result = "Pass" if marks >= 50 else "Fail"
print(result)
````

## 7️ Loops
### `for` Loop

````python
# Print numbers from 1 to 5
for i in range(1, 6):
    print(i)

# Table of 5
print(" Table of 5 ")
for i in range(1, 11):
    print(f"5 x {i} = {5 * i}")

# Loop through a string
for ch in "Python":
    print(ch, end=" ")
````

### `while` Loop

````python
count = 1
while count <= 5:
    print("Count:", count)
    count += 1

# Sum of numbers from 1 to 10
total = 0
n = 1
while n <= 10:
    total += n
    n += 1
print("Sum =", total)
````

### `break`, `continue`, and Nested Loops

````python
for i in range(1, 10):
    if i == 3:
        continue      # skip 3
    if i == 7:
        break         # stop the loop at 7
    print(i, end=" ")
print()

# Nested loop: print a pattern
for i in range(1, 5):
    print("*" * i)
````

## 8️ Data Structures
###  List : Ordered, Changeable, Duplicates Allowed

````python
fruits = ["apple", "banana", "mango"]
print(fruits)

fruits.append("orange")          # add at the end
fruits.insert(1, "grapes")       # add at index 1
fruits.remove("banana")          # remove by value
print(fruits)

print("First:", fruits[0], "| Last:", fruits[-1])
print("Slice:", fruits[1:3])
print("Length:", len(fruits))

fruits.sort()
print("Sorted:", fruits)

for f in fruits:
    print("-", f)

# List comprehension
squares = [n ** 2 for n in range(1, 6)]
print("Squares:", squares)
````

###  Tuple:  Ordered, **Unchangeable**

````python
point = (10, 20, 30)
print(point, type(point))
print("Index 1:", point[1])
print("Count of 10:", point.count(10))
print("Index of 30:", point.index(30))

# Unpacking
x, y, z = point
print(x, y, z)

# point[0] = 5   #  Error: a tuple cannot be changed
````

###  Set: Unordered, **Unique** Items

````python
nums = {1, 2, 3, 3, 2, 1}
print(nums)                      # duplicates are removed

nums.add(4)
nums.discard(1)
print(nums)

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print("Union:", A | B)
print("Intersection:", A & B)
print("Difference:", A - B)

# Remove duplicates from a list
data = [1, 2, 2, 3, 3, 3]
print(list(set(data)))
````

###  Dictionary :-> Key : Value Pairs

````python
student = {
    "name": "Ahmed",
    "age": 19,
    "course": "Python"
}
print(student)
print("Name:", student["name"])
print("Age:", student.get("age"))

student["age"] = 20              # update the value
student["city"] = "Islamabad"    # add a new key
del student["course"]            # delete a key
print(student)

print("Keys:", list(student.keys()))
print("Values:", list(student.values()))

for key, value in student.items():
    print(key, "->", value)
````

## 9️ Beginner Practice Problems

**Problem 1:** Find the sum, difference, product, and division of two numbers.

````python
a, b = 12, 5
print("Sum:", a + b)
print("Difference:", a - b)
print("Product:", a * b)
print("Division:", a / b)
````

**Problem 2:** Check whether a number is positive, negative, or zero.

````python
n = -8
if n > 0:
    print("Positive")
elif n < 0:
    print("Negative")
else:
    print("Zero")
````

**Problem 3:** Print the even numbers from 1 to 20.

````python
for i in range(1, 21):
    if i % 2 == 0:
        print(i, end=" ")
````

**Problem 4:** Find the factorial of a number.

````python
n = 5
fact = 1
for i in range(1, n + 1):
    fact *= i
print(f"{n}! =", fact)
````

**Problem 5:** Reverse a string and check whether it is a palindrome.

````python
word = "madam"
reversed_word = word[::-1]
print("Reverse:", reversed_word)
print("its Palindrome " if word == reversed_word else "Not Palindrome ")
````

**Problem 6:** Find the largest and smallest number in a list.

````python
numbers = [45, 12, 89, 3, 67]
print("Max:", max(numbers))
print("Min:", min(numbers))
print("Average:", sum(numbers) / len(numbers))
````

**Problem 7:** Print the first 10 numbers of the Fibonacci series.

````python
a, b = 0, 1
for _ in range(10):
    print(a, end=" ")
    a, b = b, a + b
````

**Problem 8:** Count how many times each word appears in a sentence using a dictionary.

````python
sentence = "python is easy and python is fun"
count = {}
for word in sentence.split():
    count[word] = count.get(word, 0) + 1
print(count)
````

**Problem 9:** Create a simple calculator using a function.

````python
def calculator(a, b, op):
    if op == "+":
        return a + b
    elif op == "-":
        return a - b
    elif op == "*":
        return a * b
    elif op == "/":
        return a / b if b != 0 else "Zero se divide nahi ho sakta"
    else:
        return "Invalid operator"

print(calculator(10, 5, "+"))
print(calculator(10, 5, "/"))
print(calculator(10, 0, "/"))
````

**Problem 10:** Check whether a number is prime.

````python
n = 29
is_prime = n > 1
for i in range(2, int(n ** 0.5) + 1):
    if n % i == 0:
        is_prime = False
        break
print(n, "is prime" if is_prime else "is not prime")
````
