## 1. What is OOP?

**Object-Oriented Programming (OOP)** is a way of writing programs using **objects**. An object is something that has:
- **Data** (also called *attributes*): what it *has*. Example: a car has a color and a speed.
- **Actions** (also called *methods*): what it can *do*. Example: a car can start and stop.

**Why do we use OOP?**
- Code is organized and easy to read.
- We can **reuse** code.
- It is easier to fix and update big programs.
- It models real life (bank, student, car, etc.).

**Real life example:** A *blueprint* of a house is a **class**. The real houses built from that blueprint are **objects**.

## 2. Classes and Objects

- A **class** is a blueprint (a plan).
- An **object** is a real thing made from that class. It is also called an **instance**.

````python
class Dog:
    # A simple class with no data yet
    pass

dog1 = Dog()   # dog1 is an object of class Dog
dog2 = Dog()   # dog2 is another object

print(type(dog1))
print(dog1 is dog2)   # False: they are two different objects
````

## 3. Constructors (`__init__`)

A **constructor** is a special method that runs **automatically** when we create an object. In Python it is called `__init__`.

We use it to set the starting values (attributes) of the object.

- `self` means "this object itself". It must be the first parameter of every method.

````python
class Student:
    def __init__(self, name, age):   # constructor
        self.name = name              # attribute
        self.age = age                # attribute

s1 = Student("Ali", 20)
s2 = Student("Sara", 22)

print(s1.name, s1.age)
print(s2.name, s2.age)
````

## 4. Methods

A **method** is a function that is written inside a class. It describes what an object can **do**.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def show_info(self):                 # a method
        print(f"Name: {self.name}, Marks: {self.marks}")

    def is_pass(self):                   # a method that returns a value
        return self.marks >= 50

s1 = Student("Ali", 65)
s1.show_info()
print("Passed?", s1.is_pass())
````

## 5. Encapsulation

**Encapsulation** means keeping data **safe** inside the class, and allowing changes only through methods.

In Python we show that a value is private by using **underscores**:
- `name`: public (anyone can use it)
- `_name`: protected (please do not touch, only for the class and its children)
- `__name`: private (Python hides it from outside)

**Why is it useful?** In a bank, nobody should be able to write `account.balance = 1000000` directly. Money must change only through `deposit()` and `withdraw()`.

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance          # private attribute

    def get_balance(self):                # getter (safe way to read)
        return self.__balance

    def deposit(self, amount):            # safe way to change
        if amount > 0:
            self.__balance += amount

acc = BankAccount(1000)
acc.deposit(500)
print(acc.get_balance())      # 1500

# print(acc.__balance)        # This would give an error (AttributeError)
````

## 6. Inheritance

**Inheritance** means a new class (**child**) can take the features of an old class (**parent**). This saves us from writing the same code again.

Real life: A `Dog` is an `Animal`. It gets all animal features and can add its own.

````python
class Animal:                              # parent class
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating.")

class Dog(Animal):                         # child class
    def bark(self):
        print(f"{self.name} says Woof!")

d = Dog("Tommy")
d.eat()     # inherited from Animal
d.bark()    # its own method
````

## 7. Polymorphism

**Polymorphism** means "many forms". Different classes can have the **same method name**, but each one does the work in its **own way**.

````python
class Cat:
    def speak(self):
        return "Meow"

class Cow:
    def speak(self):
        return "Moo"

for animal in [Cat(), Cow()]:
    print(animal.speak())      # same method name, different result
````

## 8. Abstraction

**Abstraction** means showing **only what is needed** and hiding the details.

In Python we use an **abstract class** (from the `abc` module). It is a class that:
- cannot be used to make objects directly,
- forces every child class to write certain methods.

Real life: When you drive a car, you use the steering wheel. You do not need to know how the engine works inside.

````python
from abc import ABC, abstractmethod

class Shape(ABC):                    # abstract class
    @abstractmethod
    def area(self):                  # child classes MUST write this
        pass

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h

class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r * self.r

for shape in [Rectangle(4, 5), Circle(3)]:
    print(type(shape).__name__, "area =", shape.area())

# s = Shape()   # Error! We cannot create an object of an abstract class
````

## 9. Clean Code Principles

Clean code is code that is **easy to read, easy to understand and easy to change**.

| Principle | Meaning | Example |
|---|---|---|
| **Meaningful names** | Names should explain themselves | `total_price`, not `tp` |
| **Small functions** | One function does **one** job | `deposit()` only deposits |
| **DRY** (Don't Repeat Yourself) | Do not copy-paste code | Use a helper method |
| **Single Responsibility** | One class has **one** reason to change | `Transaction` only stores a record |
| **Comments and docstrings** | Explain *why*, not the obvious | `"""Add money."""` |
| **Handle errors** | Use exceptions with clear messages | `InsufficientFundsError` |
| **Consistent style** | Follow PEP 8 (snake_case, 4 spaces) | `get_balance()` |

We will follow all of these in our Banking System below.

## 10. Banking System (Full Project)

### Requirements
- Create an account (Savings or Current)
- Deposit money
- Withdraw money (with rules)
- Show balance
- Keep and show **transaction history**
- Transfer money between accounts (bonus)

### Design plan
| Class | Job |
|---|---|
| `Transaction` | Stores one record (deposit, withdraw, etc.) |
| `Account` (abstract) | Common features for all accounts |
| `SavingsAccount` | Account with a **minimum balance** and **interest** |
| `CurrentAccount` | Account with an **overdraft limit** |
| `Bank` | Creates and manages accounts |
| Custom errors | `InvalidAmountError`, `InsufficientFundsError`, `AccountNotFoundError` |

### Step 1: Custom errors

````python
class BankError(Exception):
    """Base class for all banking errors."""

class InvalidAmountError(BankError):
    """Raised when the amount is zero or negative."""

class InsufficientFundsError(BankError):
    """Raised when the account does not have enough money."""

class AccountNotFoundError(BankError):
    """Raised when an account number does not exist."""
````

### Step 2: The `Transaction` class
This class only keeps information about **one** transaction.

````python
from datetime import datetime

class Transaction:
    """A record of one bank transaction."""

    def __init__(self, transaction_type, amount, balance_after):
        self.transaction_type = transaction_type   # e.g. "Deposit"
        self.amount = amount
        self.balance_after = balance_after
        self.timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    def __str__(self):
        return (f"{self.timestamp} | {self.transaction_type:<13} | "
                f"Amount: {self.amount:>10.2f} | Balance: {self.balance_after:>10.2f}")
````

### Step 3: The abstract `Account` class

- The balance is **private** (encapsulation).
- `withdraw()` rules are different for each account type, so we make `_can_withdraw()` **abstract** (abstraction).
- `deposit()` and `withdraw()` are written once here and reused by all children (inheritance, DRY).

````python
from abc import ABC, abstractmethod

class Account(ABC):
    """Base class for all bank accounts."""

    def __init__(self, account_number, holder_name, opening_balance=0):
        if opening_balance < 0:
            raise InvalidAmountError("Opening balance cannot be negative.")
        self._account_number = account_number
        self._holder_name = holder_name
        self.__balance = 0                     # private
        self._transactions = []                # list of Transaction objects

        if opening_balance > 0:
            self.deposit(opening_balance)

    # ---------- Read-only properties ----------
    @property
    def account_number(self):
        return self._account_number

    @property
    def holder_name(self):
        return self._holder_name

    @property
    def balance(self):
        return self.__balance

    # ---------- Main actions ----------
    def deposit(self, amount):
        """Add money to the account."""
        self._check_amount(amount)
        self.__balance += amount
        self._record("Deposit", amount)

    def withdraw(self, amount):
        """Take money out of the account (rules depend on account type)."""
        self._check_amount(amount)
        if not self._can_withdraw(amount):
            raise InsufficientFundsError(
                f"Cannot withdraw {amount:.2f}. Current balance: {self.__balance:.2f}")
        self.__balance -= amount
        self._record("Withdraw", amount)

    def get_history(self):
        """Return a copy of the transaction list."""
        return list(self._transactions)

    def print_history(self):
        print(f"--- Transaction History: {self._holder_name} ({self._account_number}) ---")
        if not self._transactions:
            print("No transactions yet.")
        for t in self._transactions:
            print(t)

    # ---------- Abstract rule: each child decides ----------
    @abstractmethod
    def _can_withdraw(self, amount):
        """Return True if the withdrawal is allowed."""

    @abstractmethod
    def account_type(self):
        """Return the name of the account type."""

    # ---------- Helper methods (protected) ----------
    def _check_amount(self, amount):
        if not isinstance(amount, (int, float)) or amount <= 0:
            raise InvalidAmountError("Amount must be a positive number.")

    def _record(self, transaction_type, amount):
        self._transactions.append(Transaction(transaction_type, amount, self.__balance))

    def _add_to_balance_with_record(self, transaction_type, amount):
        """Used by child classes for special transactions such as interest."""
        self.__balance += amount
        self._record(transaction_type, amount)

    def __str__(self):
        return (f"[{self.account_type()}] {self._account_number} | "
                f"{self._holder_name} | Balance: {self.__balance:.2f}")
````

### Step 4: Child classes (inheritance + polymorphism)

- **SavingsAccount:** balance must never go below the **minimum balance**. It can earn **interest**.
- **CurrentAccount:** can go below zero up to an **overdraft limit**.

Both classes have `_can_withdraw()` but each one works in its own way. That is **polymorphism**.

````python
class SavingsAccount(Account):
    def __init__(self, account_number, holder_name, opening_balance=0,
                 minimum_balance=500, interest_rate=0.05):
        self.minimum_balance = minimum_balance
        self.interest_rate = interest_rate
        super().__init__(account_number, holder_name, opening_balance)

    def _can_withdraw(self, amount):
        return (self.balance - amount) >= self.minimum_balance

    def account_type(self):
        return "Savings"

    def add_interest(self):
        """Add yearly interest to the balance."""
        interest = self.balance * self.interest_rate
        if interest > 0:
            self._add_to_balance_with_record("Interest", interest)


class CurrentAccount(Account):
    def __init__(self, account_number, holder_name, opening_balance=0,
                 overdraft_limit=1000):
        self.overdraft_limit = overdraft_limit
        super().__init__(account_number, holder_name, opening_balance)

    def _can_withdraw(self, amount):
        return (self.balance - amount) >= -self.overdraft_limit

    def account_type(self):
        return "Current"
````

### Step 5: The `Bank` class

The `Bank` class **manages** many accounts. It creates accounts, finds them, and transfers money.

````python
class Bank:
    """Manages all accounts."""

    def __init__(self, name):
        self.name = name
        self._accounts = {}          # account_number -> Account
        self._next_number = 1001     # auto-generated account numbers

    def create_account(self, account_type, holder_name, opening_balance=0):
        """Create a 'savings' or 'current' account and return it."""
        number = f"ACC{self._next_number}"
        kind = account_type.lower()

        if kind == "savings":
            account = SavingsAccount(number, holder_name, opening_balance)
        elif kind == "current":
            account = CurrentAccount(number, holder_name, opening_balance)
        else:
            raise ValueError("Account type must be 'savings' or 'current'.")

        self._accounts[number] = account
        self._next_number += 1
        return account

    def find_account(self, account_number):
        if account_number not in self._accounts:
            raise AccountNotFoundError(f"Account {account_number} not found.")
        return self._accounts[account_number]

    def transfer(self, from_number, to_number, amount):
        """Move money from one account to another."""
        source = self.find_account(from_number)
        target = self.find_account(to_number)
        source.withdraw(amount)      # raises an error if not allowed
        target.deposit(amount)

    def show_all_accounts(self):
        print(f"=== {self.name}: All Accounts ===")
        for account in self._accounts.values():
            print(account)
````

### Step 6: Let us test the system

````python
bank = Bank("Python National Bank")

# 1. Create accounts
ali = bank.create_account("savings", "Ali Khan", 5000)
sara = bank.create_account("current", "Sara Ahmed", 2000)

bank.show_all_accounts()
````

````python
# 2. Deposit and withdraw
ali.deposit(1500)
ali.withdraw(1000)
sara.withdraw(2500)      # allowed: current account can use overdraft

print(ali)
print(sara)
````

````python
# 3. Add interest to the savings account
ali.add_interest()
print(ali)
````

````python
# 4. Transfer money
bank.transfer(ali.account_number, sara.account_number, 1000)
print(ali)
print(sara)
````

````python
# 5. Transaction history
ali.print_history()
print()
sara.print_history()
````

### Step 7: Testing errors (they must be handled safely)

````python
tests = [
    ("Withdraw too much from savings", lambda: ali.withdraw(999999)),
    ("Deposit a negative amount",      lambda: ali.deposit(-50)),
    ("Overdraft limit crossed",        lambda: sara.withdraw(5000)),
    ("Unknown account",                lambda: bank.find_account("ACC9999")),
    ("Wrong account type",             lambda: bank.create_account("gold", "Test")),
]

for title, action in tests:
    try:
        action()
    except (BankError, ValueError) as error:
        print(f"{title}: {type(error).__name__} -> {error}")
````

### Picture version
Run the next cell to draw the diagram with `matplotlib`.

````python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

fig, ax = plt.subplots(figsize=(12, 8))
ax.set_xlim(0, 12)
ax.set_ylim(0, 8)
ax.axis("off")

def draw_class(x, y, w, h, title, attributes, methods, color):
    ax.add_patch(FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.02",
                                fc="white", ec="black", lw=1.5))
    ax.add_patch(FancyBboxPatch((x, y + h - 0.6), w, 0.6, boxstyle="round,pad=0.02",
                                fc=color, ec="black", lw=1.5))
    ax.text(x + w / 2, y + h - 0.3, title, ha="center", va="center",
            fontsize=10, fontweight="bold")
    text_y = y + h - 0.85
    for line in attributes:
        ax.text(x + 0.1, text_y, line, fontsize=8, va="center")
        text_y -= 0.3
    ax.plot([x, x + w], [text_y + 0.1, text_y + 0.1], color="black", lw=0.8)
    text_y -= 0.2
    for line in methods:
        ax.text(x + 0.1, text_y, line, fontsize=8, va="center")
        text_y -= 0.3

def arrow(x1, y1, x2, y2, label="", style="->"):
    ax.annotate("", xy=(x2, y2), xytext=(x1, y1),
                arrowprops=dict(arrowstyle=style, lw=1.5))
    if label:
        ax.text((x1 + x2) / 2 + 0.1, (y1 + y2) / 2, label, fontsize=8, style="italic")

# Account (abstract)
draw_class(4, 4.6, 4, 3.0, "<<abstract>> Account",
           ["- __balance", "# _account_number", "# _holder_name", "# _transactions"],
           ["+ deposit(amount)", "+ withdraw(amount)", "+ print_history()",
            "# _can_withdraw()*", "+ account_type()*"], "#cfe8ff")

# Children
draw_class(0.3, 1.6, 3.7, 2.1, "SavingsAccount",
           ["+ minimum_balance", "+ interest_rate"],
           ["# _can_withdraw()", "+ add_interest()"], "#d5f5d5")
draw_class(8, 1.6, 3.7, 2.1, "CurrentAccount",
           ["+ overdraft_limit"],
           ["# _can_withdraw()", "+ account_type()"], "#d5f5d5")

# Transaction
draw_class(4.4, 0.2, 3.2, 1.9, "Transaction",
           ["+ transaction_type", "+ amount", "+ balance_after", "+ timestamp"],
           [], "#ffe9c2")

# Bank
draw_class(8.6, 5.0, 3.2, 2.4, "Bank",
           ["+ name", "- _accounts"],
           ["+ create_account()", "+ find_account()", "+ transfer()"], "#f5d5f5")

# Inheritance arrows (child -> parent)
arrow(2.2, 3.7, 4.6, 4.6, "inherits", "-|>")
arrow(9.8, 3.7, 7.4, 4.6, "inherits", "-|>")
# Bank has many Accounts
arrow(8.6, 6.2, 8.0, 6.2, "1..*", "->")
# Account has many Transactions
arrow(6.0, 4.6, 6.0, 2.1, "1..*", "->")

ax.set_title("Banking System: Class Diagram", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
````

## 12. Design Documentation

### 12.1 Purpose
This application is a small **Banking System**. It lets a user create accounts, deposit and withdraw money, transfer money, and see the transaction history.

### 12.2 How each OOP concept is used

| OOP Concept | Where we used it | Why it helps |
|---|---|---|
| **Class and Object** | `Account`, `Bank`, `Transaction`; `ali`, `sara` | Models real bank things |
| **Constructor** | `__init__` in every class | Sets starting values safely |
| **Methods** | `deposit()`, `withdraw()`, `transfer()` | Actions the objects can do |
| **Encapsulation** | `__balance` is private; only `deposit`/`withdraw` change it | Nobody can cheat the balance |
| **Inheritance** | `SavingsAccount` and `CurrentAccount` extend `Account` | Shared code is written once |
| **Polymorphism** | Each account has its own `_can_withdraw()` | Same call, different rule |
| **Abstraction** | `Account` is an abstract class | Hides details, forces rules in children |

### 12.3 Clean architecture and responsibilities
Each class has **one clear job** (Single Responsibility Principle):

- `Transaction`: only stores a record.
- `Account` and its children: handle the money rules of **one** account.
- `Bank`: manages **many** accounts and transfers. It does not know the withdrawal rules; it just asks the account.
- Custom errors: describe what went wrong in clear words.

**Dependency direction:** `Bank` depends on `Account`. `Account` depends on `Transaction`. Nothing depends back on `Bank`. This keeps the design simple and easy to test.

### 12.4 Business rules
1. Every amount must be a **positive number**.
2. **Savings:** balance cannot go below the minimum balance (500).
3. **Current:** balance cannot go below the overdraft limit (-1000).
4. Every deposit, withdrawal and interest payment is saved in the history.
5. A transfer is a withdrawal from one account followed by a deposit to another.

### 12.5 Error handling
| Error | When it happens |
|---|---|
| `InvalidAmountError` | Amount is zero, negative or not a number |
| `InsufficientFundsError` | Withdrawal breaks the account rule |
| `AccountNotFoundError` | Account number does not exist |
